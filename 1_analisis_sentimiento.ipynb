```python
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

train_x_vector = tfidf.fit_transform(train_x)
test_x_vector = tfidf.transform(test_x)

print("\nForma de la matriz de entrenamiento:")
print(train_x_vector.shape)

print("\nForma de la matriz de prueba:")
print(test_x_vector.shape)
```

# 5. Crear los Modelos

```python
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix
```

```python
# 5. Definir los modelos a evaluar
modelos = {
    "SVM": SVC(kernel="linear"),
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "GaussianNB": GaussianNB(),
    "Decision Tree": DecisionTreeClassifier(random_state=42)
}
```

```python
# 6. Entrenar y evaluar cada modelo

resultados = []

for nombre, modelo in modelos.items():

    print("\n==========================================")
    print(f"Modelo: {nombre}")
    print("==========================================")

    # GaussianNB necesita matrices densas.
    # Los demás modelos pueden trabajar con matrices dispersas.
    if nombre == "GaussianNB":
        modelo.fit(train_x_vector.toarray(), train_y)
        predicciones = modelo.predict(test_x_vector.toarray())
    else:
        modelo.fit(train_x_vector, train_y)
        predicciones = modelo.predict(test_x_vector)

    accuracy = accuracy_score(test_y, predicciones)
    f1_positivo = f1_score(
        test_y,
        predicciones,
        pos_label="positive"
    )
    f1_negativo = f1_score(
        test_y,
        predicciones,
        pos_label="negative"
    )

    resultados.append({
        "Modelo": nombre,
        "Accuracy": accuracy,
        "F1 positivo": f1_positivo,
        "F1 negativo": f1_negativo
    })

    print("\nAccuracy:")
    print(accuracy)

    print("\nF1 positivo:")
    print(f1_positivo)

    print("\nF1 negativo:")
    print(f1_negativo)

    print("\nReporte de clasificación:")
    print(classification_report(test_y, predicciones))

    print("\nMatriz de confusión:")
    print(confusion_matrix(
        test_y,
        predicciones,
        labels=["positive", "negative"]
    ))
```

```python
#Comparar resultados de los modelos
df_resultados = pd.DataFrame(resultados)

print("\n==========================================")
print("Comparación final de modelos")
print("==========================================")
print(df_resultados)

mejor_modelo = df_resultados.sort_values(
    by="Accuracy",
    ascending=False
).iloc[0]

print("\nMejor modelo según Accuracy:")
print(mejor_modelo)
```

```python
# Comparar resultados de los modelos SVM vs Logistic Regression
from IPython.display import display
import pandas as pd

frases = [
    "A good movie",
    "An excellent movie",
    "I did not like this movie at all",
    "This movie was boring and terrible",
    "I loved the story and the acting was great"
]

# Vectorizar frases nuevas
frases_vector = tfidf.transform(frases)

# Entrenar modelo SVM
modelo_svm = SVC(kernel="linear")
modelo_svm.fit(train_x_vector, train_y)

# Entrenar modelo Logistic Regression
modelo_logistic = LogisticRegression(max_iter=1000)
modelo_logistic.fit(train_x_vector, train_y)

# Predicciones
predicciones_svm = modelo_svm.predict(frases_vector)
predicciones_logistic = modelo_logistic.predict(frases_vector)

# Crear tabla comparativa
tabla_predicciones = pd.DataFrame({
    "Frase": frases,
    "Predicción SVM": predicciones_svm,
    "Predicción Logistic Regression": predicciones_logistic
})

print("\n==========================================")
print("Comparación de predicciones por modelo")
print("==========================================")
display(tabla_predicciones)
```

# CONCLUSION

Después de entrenar y evaluar los modelos SVM, Logistic Regression, GaussianNB y Decision Tree, se observó que el modelo con mejor desempeño fue Logistic Regression, con un accuracy de 0.870606. Este resultado fue ligeramente superior al obtenido por SVM, que alcanzó un accuracy de 0.869394.

Aunque la diferencia entre Logistic Regression y SVM fue pequeña, Logistic Regression obtuvo el mejor resultado general y también presentó un F1-score positivo de 0.873406 y un F1-score negativo de 0.867679. Esto indica que el modelo logró clasificar de manera equilibrada tanto reseñas positivas como negativas.

Por otro lado, GaussianNB obtuvo un rendimiento menor, con un accuracy de 0.771212, mientras que Decision Tree fue el modelo con el desempeño más bajo, con un accuracy de 0.714242. Esto sugiere que estos modelos no fueron tan adecuados para este problema de clasificación de texto usando TF-IDF.

Asì, aunque SVM era una buena opción inicial para el análisis de sentimientos, en esta prueba Logistic Regression fue el mejor modelo, ya que obtuvo el accuracy más alto y mantuvo un buen equilibrio entre las métricas de ambas clases. Por esta razón, se seleccionó Logistic Regression como el modelo final para realizar predicciones con nuevas reseñas.

Por último, para comprobar el comportamiento de los dos mejores modelos, se probaron varias frases nuevas con SVM y Logistic Regression. Los resultados se organizaron en una tabla comparativa, donde se muestra la predicción realizada por cada modelo para la misma frase.

Esta comparación permite observar si ambos modelos clasifican las reseñas de manera similar o si existen diferencias en sus predicciones. En este caso, como Logistic Regression obtuvo el mejor accuracy general, sus predicciones se consideran las más confiables dentro de los modelos evaluados.
